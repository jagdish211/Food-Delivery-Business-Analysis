
## Missing Values Handling

During the data cleaning process, missing values were checked using:

```python
df.isnull().sum()
```

The dataset contains missing values in a few columns. These missing values were not blindly filled because they have a valid reason.

### Missing Values Found

| Column                         | Missing Values | Reason                                                        |
| ------------------------------ | -------------: | ------------------------------------------------------------- |
| `cancellation_reason`          |         49,132 | Cancellation reason is not applicable to non-cancelled orders |
| `customer_rating`              |            868 | These orders were cancelled                                   |
| `late_delivery`                |            868 | Cancelled orders were not delivered                           |
| `actual_delivery_time_minutes` |            868 | No actual delivery time exists for cancelled orders           |
| `tip_amount`                   |            868 | No tip was recorded for cancelled orders                      |

### Handling Approach

* `cancellation_reason` → Kept as missing because it is not applicable to non-cancelled orders.
* `customer_rating` → Kept as missing because cancelled orders have no delivery rating.
* `late_delivery` → Kept as missing because cancelled orders were not delivered.
* `actual_delivery_time_minutes` → Kept as missing because delivery did not take place.
* `tip_amount` → Missing values can be replaced with `0` because no tip was recorded.

The 868 missing values in the delivery-related columns were confirmed to belong to **Cancelled orders**.

Therefore, these missing values are considered **meaningful missing values rather than data errors**.


## Delivery Delay -> Estimated vs actual delivery time

````python
df['delivery_delay_minutes'] = (
    df['actual_delivery_time_minutes'] -
    df['estimated_delivery_time_minutes']
)
````

## Order Value Category -> Low / Medium / High order

````python
df['order_value_category'] = pd.cut(
    df['order_total'],
    bins=[0, 300, 600, 1000, np.inf],
    labels=['Low', 'Medium', 'High', 'Very High']
)
````

## Distance Category

````python
df['distance_category'] = pd.cut(
    df['distance_km'],
    bins=[0, 3, 7, 12, np.inf],
    labels=['Short', 'Medium', 'Long', 'Very Long']
)
````

## Age Group

````python
df['age_group'] = pd.cut(
    df['customer_age'],
    bins=[0, 30, 45, 60, np.inf],
    labels=['18-30', '31-45', '46-60', '60+']
)
````

## Total Fees

````python
df['total_fees'] = (
    df['tax_amount'] +
    df['service_fee'] +
    df['delivery_fee']
)
````

## Order Month

````python
df['order_month'] = df['order_date'].dt.month
````

## Order Day

````python
df['order_day'] = df['order_date'].dt.day_name()
````

## Is Late

````python
df['is_late'] = df['delivery_delay_minutes'] > 0
````
