## Q1. Which Day Receives the Highest Number of Orders?

This analysis compares the number of food delivery orders
received on each day of the week.

The objective is to identify the day with the highest order
volume and understand weekly customer demand.

```python
day_orders = df.groupby('day_of_week')['order_id'].count()

day_orders = day_orders.sort_values(ascending=False)

day_orders
plt.figure(figsize=(10, 6))

sns.barplot(
    x=day_orders.index,
    y=day_orders.values,
    hue=day_orders.index,
    palette='Set2',
    legend=False
)

plt.title(
    'Number of Orders by Day of Week',
    fontsize=18,
    fontweight='bold'
)

plt.xlabel('Day of Week')
plt.ylabel('Number of Orders')

plt.xticks(rotation=30)

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()
```

### Business Insight

The analysis shows the number of orders received on each day
of the week.

The day with the highest number of orders represents a period
of higher customer demand.

This information can help the business plan delivery partners,
restaurant capacity and promotional activities accordingly.

## Q2. Which Day Generates the Highest Revenue?

This analysis compares the total revenue generated on each day
of the week.

The objective is to identify the day that generates the highest
revenue and understand weekly revenue patterns.

```python
day_revenue = df.groupby('day_of_week')['order_total'].sum()

day_revenue = day_revenue.sort_values(ascending=False)

day_revenue

plt.figure(figsize=(10, 6))

plt.plot(
    day_revenue.index,
    day_revenue.values,
    marker='o',
    linewidth=3,
    markersize=8
)

plt.title(
    'Revenue by Day of Week',
    fontsize=18,
    fontweight='bold'
)

plt.xlabel('Day of Week')
plt.ylabel('Total Revenue (₹)')

plt.xticks(rotation=30)

plt.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
```

### Business Insight

The analysis shows how revenue varies across different days
of the week.

The day with the highest revenue represents an important
revenue period for the business.

This information can help the business plan promotions,
restaurant capacity and delivery operations accordingly.

## Q3. Which Customer Type Places the Most Orders?

This analysis compares the number of orders placed by different
customer types.

The objective is to identify which customer type contributes
the highest number of orders.

```python
customer_orders = df.groupby('customer_type')['order_id'].count()

customer_orders = customer_orders.sort_values(ascending=False)

customer_orders

plt.figure(figsize=(8, 8))

plt.pie(
    customer_orders.values,
    labels=customer_orders.index,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'width': 0.4}
)

plt.title(
    'Orders by Customer Type',
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

customer types.

an important customer segment for the business.

This information can help the business understand its customer
base and plan suitable marketing strategies.

## Q4. Which Customer Type Has the Highest Average Order Value (AOV)?

This analysis compares the average order value of different
customer types.

The objective is to identify which customer type spends the
most per order on average.

```python
customer_aov = df.groupby('customer_type')['order_total'].mean()

customer_aov = customer_aov.sort_values(ascending=False)

customer_aov
plt.figure(figsize=(10, 6))

sns.barplot(
    x=customer_aov.values,
    y=customer_aov.index,
    hue=customer_aov.index,
    palette='coolwarm',
    legend=False
)

plt.title(
    'Average Order Value by Customer Type',
    fontsize=18,
    fontweight='bold'
)

plt.xlabel('Average Order Value (₹)')
plt.ylabel('Customer Type')

plt.grid(axis='x', alpha=0.2)

plt.tight_layout()
plt.show()
```

### Business Insight

The analysis shows the average amount spent by different
customer types in each order.

The customer type with the highest AOV represents a segment
with higher average spending per order.

This information can help the business understand customer
spending behavior and design suitable offers and promotions.

## Q10. What Is the Average Delivery Time for Each Traffic Level?

This analysis compares the average actual delivery time
for different traffic levels.

The objective is to understand how traffic conditions
affect food delivery time.

```python
traffic_delivery = df.groupby('traffic_level')['actual_delivery_time_minutes'].mean()

traffic_delivery = traffic_delivery.sort_values(ascending=False)

traffic_delivery

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x='traffic_level',
    y='actual_delivery_time_minutes',
    hue='traffic_level',
    palette='Set3',
    legend=False
)

plt.title(
    'Delivery Time by Traffic Level',
    fontsize=18,
    fontweight='bold'
)

plt.xlabel('Traffic Level')
plt.ylabel('Actual Delivery Time (Minutes)')

plt.grid(axis='y', alpha=0.2)

plt.tight_layout()
plt.show()
```

### Business Insight

The analysis compares actual delivery times across different
traffic levels.

Differences in delivery time can help the business understand
the impact of traffic conditions on delivery operations.

This information can be useful for improving delivery-time
estimates and planning delivery resources.

## Q11. Which Weather Condition Has the Most Late Deliveries?

This analysis compares the number of late deliveries under
different weather conditions.

The objective is to identify the weather condition associated
with the highest number of late deliveries.

```python
weather_late = df[df['is_late'] == 1].groupby('weather')['order_id'].count()

weather_late = weather_late.sort_values(ascending=False)

weather_late

plt.figure(figsize=(10, 5))

sns.heatmap(
    weather_late.to_frame(),
    annot=True,
    fmt='g',
    cmap='YlOrRd'
)

plt.title(
    'Late Deliveries by Weather Condition',
    fontsize=18,
    fontweight='bold'
)

plt.xlabel('Metric')
plt.ylabel('Weather Condition')

plt.tight_layout()
plt.show()
```

### Business Insight

The analysis shows the number of late deliveries under
different weather conditions.

Weather conditions with a higher number of late deliveries
may create additional challenges for delivery operations.

This information can help the business plan delivery operations
and improve estimated delivery times during difficult weather.

## Q12. Does Delivery Distance Affect Delivery Time?

This analysis examines the relationship between delivery distance
and actual delivery time.

The objective is to understand whether longer delivery distances
are associated with longer delivery times.

```python
distance_time = df[
    ['distance_km', 'actual_delivery_time_minutes']
]

distance_time.head()
plt.figure(figsize=(10, 6))

sns.regplot(
    data=df,
    x='distance_km',
    y='actual_delivery_time_minutes',
    scatter_kws={'alpha': 0.4},
    line_kws={'linewidth': 2}
)

plt.title(
    'Delivery Distance vs Actual Delivery Time',
    fontsize=18,
    fontweight='bold'
)

plt.xlabel('Delivery Distance (KM)')
plt.ylabel('Actual Delivery Time (Minutes)')

plt.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

### Business Insight

The scatter plot shows the relationship between delivery
distance and actual delivery time.

If delivery time increases as distance increases, it indicates
that longer delivery distances may require more time to complete.

This information can help the business improve delivery-time
estimates and plan delivery operations more effectively.
